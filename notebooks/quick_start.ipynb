# ProbForecast Quick Start

This notebook demonstrates how to use ProbForecast.

```python
import pandas as pd
import numpy as np
import requests
import plotly.graph_objects as go
from datetime import datetime
```

```python
# Generate sample data
from src.data.dataset import TimeSeriesDataProcessor
import yaml

with open('configs/training.yaml', 'r') as f:
    config = yaml.safe_load(f)

processor = TimeSeriesDataProcessor(config)
df = processor.generate_sample_data(n_series=5, n_periods=365)
print(f'Generated {len(df)} data points')
df.head()
```

```python
# Make forecast request
api_url = 'http://localhost:8000'

request_data = {
    'series_id': 'series_000',
    'horizon': 30,
    'model_type': 'deepar',
    'quantiles': [0.1, 0.5, 0.9]
}

response = requests.post(f'{api_url}/forecast', json=request_data)
forecast_data = response.json()

print('✅ Forecast generated!')
print(f'Horizon: {len(forecast_data["forecast"])} steps')
```
